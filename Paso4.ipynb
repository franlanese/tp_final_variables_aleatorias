Por definición, la función de distribución acumulada se expresa como:

$$
F_X(x)=P(X\le x)
$$

Por lo tanto,

$$
P(X\le12)=F_X(12)
$$

$$

$$

$$

$$
F_X(12)=0.981532
$$

Por lo tanto,

$$
