## Variable aleatoria continua

Sea la variable aleatoria continua

$$
Y=\text{Precio del juego (Price)}
$$

La variable \(Y\) representa el precio de cada videojuego.

Se considera una **variable aleatoria continua**, ya que puede tomar distintos valores dentro de un intervalo, incluyendo valores decimales (por ejemplo, 3.99, 7.19 o 19.99).

Aunque en el dataset los precios se registran con dos cifras decimales, desde el punto de vista estadístico se modelan como una variable continua.
