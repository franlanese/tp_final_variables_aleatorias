## I. Construcción de la tabla de frecuencias

Sea \(X\) la variable aleatoria discreta que representa la **edad mínima requerida** (*Required Age*) para jugar un videojuego del dataset de Steam.

Para construir la tabla de frecuencias se siguieron los siguientes pasos.

### Paso 1 — Identificar los valores posibles

Se analizaron los **27075** registros del dataset y se identificaron todos los valores distintos que puede tomar la variable.

$$
R_X=\{0,3,7,12,16,18\}
$$

### Paso 2 — Calcular la frecuencia absoluta

Para cada valor del recorrido se contó la cantidad de juegos que poseen esa edad mínima requerida. A esta cantidad se la denomina **frecuencia absoluta** y se representa mediante: f<sub>i</sub>

### Paso 3 — Calcular la probabilidad puntual

La probabilidad asociada a cada valor de la variable se obtuvo dividiendo la frecuencia absoluta entre el número total de registros del dataset.

$$
p_X(x)=P(X=x)=\frac{f<sub>i</sub>}{N}
$$

donde:

- \(f<sub>i</sub>\) es la frecuencia absoluta.
- \(N = 27075\) es la cantidad total de observaciones.

**Paso 1 — Para \(x=0\):**

La frecuencia absoluta es \(f<sub>i</sub>=26479\).

$$
P(X=0)=\frac{26479}{27075}
$$

$$
P(X=0)=0.977987
$$


**Paso 2 — Para \(x=3\):**

La frecuencia absoluta es \(f<sub>i</sub>=11\).

$$
P(X=3)=\frac{11}{27075}
$$

$$
P(X=3)=0.000406
$$


**Paso 3 — Para \(x=7\):**

La frecuencia absoluta es \(f<sub>i</sub>=12\).

$$
P(X=7)=\frac{12}{27075}
$$

$$
P(X=7)=0.000443
$$


**Paso 4 — Para \(x=12\):**

La frecuencia absoluta es \(f<sub>i</sub>=73\).

$$
P(X=12)=\frac{73}{27075}
$$

$$
P(X=12)=0.002696
$$


**Paso 5 — Para \(x=16\):**

La frecuencia absoluta es \(f<sub>i</sub>=192\).

$$
P(X=16)=\frac{192}{27075}
$$

$$
P(X=16)=0.007091
$$


**Paso 6 — Para \(x=18\):**

La frecuencia absoluta es \(f<sub>i</sub>=308\).

$$
P(X=18)=\frac{308}{27075}
$$

$$
P(X=18)=0.011376
$$

## IV. Cálculos

### E[X] (Esperanza)

La esperanza matemática de una variable aleatoria discreta se calcula como:

$$
E[X] = \sum x \cdot p(x)
$$

Aplicando esto a nuestra variable:

$$
E[X] = 0 \times P(X=0) + 3 \times P(X=3) + 7 \times P(X=7) + 12 \times P(X=12) + 16 \times P(X=16) + 18 \times P(X=18)
$$

Sustituyendo los valores de probabilidad obtenidos en la tabla:

$$
E[X] = 0 \times 0.977987 + 3 \times 0.000406 + 7 \times 0.000443 + 12 \times 0.002696 + 16 \times 0.007091 + 18 \times 0.011376
$$

**Paso 1 — Término para $x=0$:** como está multiplicado por 0, este término se anula:

$$
0 \times 0.977987 = 0
$$

**Paso 2 — Término para $x=3$:**

$$
3 \times 0.000406 = 0.001218
$$

**Paso 3 — Término para $x=7$:**

$$
7 \times 0.000443 = 0.003101
$$

**Paso 4 — Término para $x=12$:**

$$
12 \times 0.002696 = 0.032352
$$

**Paso 5 — Término para $x=16$:**

$$
16 \times 0.007091 = 0.113456
$$

**Paso 6 — Término para $x=18$:**

$$
18 \times 0.011376 = 0.204768
$$

**Paso 7 — Sumar todos los términos:**

$$
E[X] = 0 + 0.001218 + 0.003101 + 0.032352 + 0.113456 + 0.204768
$$

$$
E[X] = 0.354895
$$

Por lo tanto, la **edad mínima requerida promedio** es de aproximadamente **0.355 años**, lo cual es esperable dado que la gran mayoría de los juegos no tienen restricción de edad (el 97.8% tiene edad requerida 0).

### Cálculo de E[X²]

Para calcular la varianza, primero necesitamos el segundo momento $E[X^2]$:

$$
E[X^2] = \sum x^2 \cdot p(x)
$$

Desarrollando:

$$
E[X^2] = 0^2 \times P(X=0) + 3^2 \times P(X=3) + 7^2 \times P(X=7) + 12^2 \times P(X=12) + 16^2 \times P(X=16) + 18^2 \times P(X=18)
$$

Sustituyendo los valores:

$$
E[X^2] = 0 \times 0.977987 + 9 \times 0.000406 + 49 \times 0.000443 + 144 \times 0.002696 + 256 \times 0.007091 + 324 \times 0.011376
$$

**Paso 1 — Término para $x=0$:**

$$
0^2 \times 0.977987 = 0
$$

**Paso 2 — Término para $x=3$:**

$$
3^2 \times 0.000406 = 9 \times 0.000406 = 0.003654
$$

**Paso 3 — Término para $x=7$:**

$$
7^2 \times 0.000443 = 49 \times 0.000443 = 0.021707
$$

**Paso 4 — Término para $x=12$:**

$$
12^2 \times 0.002696 = 144 \times 0.002696 = 0.388224
$$

**Paso 5 — Término para $x=16$:**

$$
16^2 \times 0.007091 = 256 \times 0.007091 = 1.815296
$$

**Paso 6 — Término para $x=18$:**

$$
18^2 \times 0.011376 = 324 \times 0.011376 = 2.912256
$$

**Paso 7 — Sumar todos los términos:**

$$
E[X^2] = 0 + 0.003654 + 0.021707 + 0.388224 + 1.815296 + 2.912256
$$

$$
E[X^2] = 5.915705
$$

### Cálculo de Var(X) (Varianza)

La varianza se calcula mediante la fórmula:

$$
Var(X) = E[X^2] - (E[X])^2
$$

**Paso 1 — Identificar los valores obtenidos:**

$$
E[X^2] = 5.915705
$$

$$
E[X] = 0.354895
$$

**Paso 2 — Calcular $(E[X])^2$:**

$$
(E[X])^2 = (0.354895)^2 = 0.125951
$$

**Paso 3 — Aplicar la fórmula:**

$$
Var(X) = 5.915705 - 0.125951
$$

$$
Var(X) = 5.789754
$$



Por lo tanto, la **varianza** de la edad mínima requerida es de **5.789754**.

# V. Construir y graficar la Función de Distribución Acumulada

La función de distribución acumulada de una variable aleatoria discreta se define como la probabilidad de que la variable tome un valor **menor o igual** que un determinado valor \(x\).

$$
F_X(x)=P(X\le x)
$$

Para calcularla, se acumulan las probabilidades obtenidas en la función de probabilidad puntual.

### Paso 1 — Para \(x=0\)

Como no existen valores menores que 0 en el recorrido de la variable:

$$
F_X(0)=P(X\le0)=P(X=0)
$$

$$
F_X(0)=0.977987
$$


### Paso 2 — Para \(x=3\)

Se acumulan las probabilidades correspondientes a los valores 0 y 3.

$$
F_X(3)=P(X\le3)
$$

$$
F_X(3)=P(X=0)+P(X=3)
$$

$$
F_X(3)=0.977987+0.000406
$$

$$
F_X(3)=0.978393
$$


### Paso 3 — Para \(x=7\)

$$
F_X(7)=P(X\le7)
$$

$$
F_X(7)=P(X=0)+P(X=3)+P(X=7)
$$

$$
F_X(7)=0.977987+0.000406+0.000443
$$

$$
F_X(7)=0.978836
$$


### Paso 4 — Para \(x=12\)

$$
F_X(12)=P(X\le12)
$$

$$
F_X(12)=P(X=0)+P(X=3)+P(X=7)+P(X=12)
$$

$$
F_X(12)=0.977987+0.000406+0.000443+0.002696
$$

$$
F_X(12)=0.981532
$$


### Paso 5 — Para \(x=16\)

$$
F_X(16)=P(X\le16)
$$

$$
F_X(16)=P(X=0)+P(X=3)+P(X=7)+P(X=12)+P(X=16)
$$

$$
F_X(16)=0.977987+0.000406+0.000443+0.002696+0.007091
$$

$$
F_X(16)=0.988623
$$


### Paso 6 — Para \(x=18\)

$$
F_X(18)=P(X\le18)
$$

$$
F_X(18)=P(X=0)+P(X=3)+P(X=7)+P(X=12)+P(X=16)+P(X=18)
$$

$$
F_X(18)=0.977987+0.000406+0.000443+0.002696+0.007091+0.011376
$$

$$
F_X(18)=0.999999\approx1
$$

Debido al redondeo de las probabilidades a seis cifras decimales, el valor final resulta ligeramente inferior a 1. Utilizando las probabilidades con toda su precisión, se obtiene exactamente:

$$
F_X(18)=1
$$

## Análisis de la distribución de la variable `required_age`

Al observar la función de probabilidad puntual de la variable edad mínima requerida (`required_age`), se aprecia que la mayor parte de la probabilidad se concentra en el valor **0**, ya que aproximadamente el **97.8 %** de los juegos no poseen una restricción de edad. Como consecuencia, la distribución presenta un pico muy pronunciado en dicho valor y probabilidades muy pequeñas para el resto de los valores (3, 7, 12, 16 y 18).

### Comparación con distribuciones teóricas

- **Distribución Binomial:**  
  No presenta una forma de campana ni una distribución aproximadamente simétrica, por lo que no se asemeja a una distribución binomial.

- **Distribución de Poisson:**  
  Aunque el valor de mayor probabilidad corresponde al menor valor de la variable, la distribución no presenta un decaimiento suave y progresivo, por lo que no puede considerarse similar a una distribución de Poisson.

- **Distribución Geométrica:**  
  La distribución comparte con la geométrica el hecho de concentrar la mayor probabilidad en el primer valor. Sin embargo, las probabilidades posteriores no decrecen de forma monótona, sino que presentan pequeños incrementos entre algunos valores, por lo que tampoco se ajustan visualmente a una distribución geométrica.

### Conclusión

La distribución de la variable `required_age` **no presenta una similitud visual clara** con ninguna de las distribuciones notables analizadas (Binomial, Poisson o Geométrica). Esto se debe a que la gran mayoría de los juegos del dataset tienen una edad mínima requerida de 0 años, generando una distribución altamente concentrada en un único valor.
